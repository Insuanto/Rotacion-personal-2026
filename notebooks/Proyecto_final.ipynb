```python
# 2) Regla del IQR para detectar outliers de forma objetiva

col = "monthly_income"

q1 = data_cruda[col].quantile(0.25)
q3 = data_cruda[col].quantile(0.75)
iqr = q3 - q1
limite_inf = q1 - 1.5 * iqr
limite_sup = q3 + 1.5 * iqr

outliers = data_cruda[(data_cruda[col] < limite_inf) | (data_cruda[col] > limite_sup)]
print(f"Rango esperado: [{limite_inf:.2f}, {limite_sup:.2f}]")
print(f"Outliers detectados: {len(outliers)} ({len(outliers)/len(data_cruda)*100:.1f}%)")
outliers.head()
```

### 3.1.1 Análisis Descriptivo de los Outliers

```python
# Estadísticas descriptivas para los outliers de 'monthly_income'
print("Estadísticas descriptivas de los 78 outliers en 'monthly_income':")
display(outliers.describe())
```

### 3.1.2 Distribución de variables categóricas en los Outliers

Podemos examinar las variables categóricas para ver si hay patrones específicos entre estos empleados con altos ingresos.

```python
# Análisis para variables categóricas en los outliers
print("Distribución de 'job_role' entre los outliers:")
display(outliers['job_role'].value_counts(normalize=True).head()) # Top 5 roles

print("\nDistribución de 'department' entre los outliers:")
display(outliers['department'].value_counts(normalize=True).head()) # Top 5 departamentos

print("\nDistribución de 'job_level' entre los outliers:")
display(outliers['job_level'].value_counts(normalize=True).head())
```

En conclusión, los empleados con ingresos mensuales atípicos son, en su mayoría, empleados con mayor antigüedad y experiencia, que ocupan puestos de alta jerarquía (principalmente gerenciales o especializados) en el departamento de Ingeniería. Este perfil es muy consistente y nos indica dónde se concentra el talento mejor remunerado de la empresa.
Esto justificaría los outliers detectados.

# 4. Análisis de datos (EDA)

Reglas de esta etapa:
- **Mínimo 3 y máximo 5 gráficas** en todo el EDA. Elegí las que mejor respondan TUS preguntas — más gráficos no es mejor análisis.
- Cada gráfica: título descriptivo (que diga *qué* muestra), ejes rotulados **con unidades**, y una interpretación escrita debajo.
- Buenas prácticas: ordenar barras por valor, no usar torta con más de ~5 categorías, no abusar de colores, resaltar solo lo que querés comunicar.

```python
# --- Bivariado: cuantitativa × cuantitativa ---

correlacion = df[["overtime_hours_per_week", "burnout_score"]].corr()
print(correlacion)


plt.figure(figsize=(9, 7))
plt.scatter(df["overtime_hours_per_week"], df["burnout_score"], alpha=0.5)
plt.title("Relación entre Horas Extras Semanales y Puntuación de Agotamiento")
plt.xlabel("Horas Extras por Semana")
plt.ylabel("Puntuación de Agotamiento")
plt.tight_layout()
plt.show()
```

Como se observa en el diagrama y confirma el coeficiente de correlación de 0,56, existe una relación positiva moderada entre estas dos variables. Esto significa que, en general, a medida que aumenta el número de horas extras semanales, la puntuación de agotamiento también tiende a aumentar. (Responde pregunta 1)


```python
# --- Bivariado: cualitativa × cuantitativa ---

resumen_por_grupo = (
    df
    .groupby("work_mode")["work_life_balance_score"]
    .agg(["count", "mean", "median", "std"])
    .round(2)
    .sort_values("median", ascending=False)
)
print(resumen_por_grupo)

# Boxplot por grupo
df.boxplot(column="work_life_balance_score", by="work_mode", figsize=(9, 7))
plt.title("Distribución de Equilibrio Vida-Trabajo por Modalidad de Trabajo")
plt.xlabel("Modalidad de Trabajo")
plt.ylabel("Puntuación de Equilibrio Vida-Trabajo")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```

El diagrama de caja muestra que la distribución de las puntuaciones de equilibrio entre la vida laboral y personal es generalmente más alta para los trabajadores remotos y progresivamente más baja para los trabajadores híbridos y presenciales. La dispersión de las puntuaciones (indicada por la caja y los bigotes) parece bastante similar en los tres grupos, lo que sugiere una variabilidad comparable en el equilibrio entre la vida laboral y personal dentro de cada modalidad. (Responde pregunta 2)


```python
# --- Bivariado: cualitativa × cuantitativa ---

resumen_por_grupo = (
    df
    .groupby("uses_ai_tools_at_work")["age"]
    .agg(["count", "mean", "median", "std"])
    .round(2)
    .sort_values("median", ascending=False)
)
print(resumen_por_grupo)


# Boxplot por grupo
df.boxplot(column="age", by="uses_ai_tools_at_work", figsize=(9, 7))
plt.title("Distribución de edad por uso de IA en el trabajo")
plt.xlabel("Uso de IA en el trabajo")
plt.ylabel("Edad de empleados")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()

```

Tanto el grupo «False» (sin herramientas de IA) como el grupo «True» (con herramientas de IA) presentan distribuciones de edad muy similares. La edad media en ambos grupos es de aproximadamente 37,12 años, y la mediana es de 37 años.
El diagrama de caja respalda visualmente estos hallazgos.
Esto sugiere que no existe una diferencia significativa en la distribución de edad entre los empleados que utilizan herramientas de IA en el trabajo y aquellos que no lo hacen.
En conclusión, según este análisis, la edad del empleado no parece ser un factor determinante en el uso de herramientas de IA en el trabajo. (Responde pergunta 3)
